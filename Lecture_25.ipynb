```python
class Calc:
  def addData(self,num1,num2):
    return num1+num2
  def minData(self,num1,num2):
    return num1-num2



obj=Calc()
print(obj.addData(10,20))
print(obj.minData(30,20))
```

```python
class Laptop():
  def __init__(self,brand,model,price):
    self.brand=brand
    self.model=model
    self.price=price
  def Apply_discount(self):
    print(f"Brand Name : {self.brand} Model Name : {self.model} Price : {self.price}")
    print(f"Discount Price : {self.price*0.10}")
    print(f"Total Price : {self.price-self.price*0.10}")

laptop1=Laptop("Dell","Inspiron",50000)
laptop2=Laptop("HP","Pavilion",30000)
laptop1.Apply_discount()
laptop2.Apply_discount()
```

```python
class MobilePhone:
  mobile_count=0
  def __init__(self,PhoneName,PhonePrice):
    self.PhoneName = PhoneName # Store PhoneName as an instance attribute
    self.PhonePrice = PhonePrice # Store PhonePrice as an instance attribute
    MobilePhone.mobile_count+=1
  def showDetails(self):
    print(f"Phone Name is {self.PhoneName} Price {self.PhonePrice}")

phone1=MobilePhone("Vivo 19",20000)
phone2=MobilePhone("Samsung",10000)
print(MobilePhone.mobile_count)
phone1.showDetails()
phone2.showDetails()
    #pass

```

```python
class Employee:
  def __init__(self,name,id, salery):
    self.name=name
    self.id=id
    self.salery=salery

  def display(self):
    print(f"Employee Name : {self.name}")
    print(f"Employee Id : {self.id}")
    print(f"Employee Salery : {self.salery}")
employee=Employee("Adil",101,10000)
employee.display()
```
